```python
nome = input("Digite seu nome: ").strip().title()
print(nome)
```

As células Python são acumulativas, e reconhecem as variáveis das célular anteriores.

Observe, por exemplo a célula abaixo:

```python
num = 5
print(f"O valor da variável num é: {num}.")
```

O valor de num é claramente 5. Porém, veja na próxima célula e diga qual será o valor da variável result:

```python
result = 15 + num
print(result)
```

O valor de result é 20 pois está somando 15 com num, que tinha o valor atribuído de 5.
